**Operations on XML File**

```python
import xml.etree.ElementTree as ET

tree = ET.parse('inventory.xml')
root = tree.getroot()

xml_data = []
for item in root.findall('item'):
    entry = {}
    for child in item:
        entry[child.tag] = child.text
    xml_data.append(entry)

xml_data[:5]
```

```python
len(xml_data)
```

```python
stock_per_category_xml = {}
for entry in xml_data:
    cat = entry['category']
    stock = int(entry['stock'])
    stock_per_category_xml[cat] = stock_per_category_xml.get(cat, 0) + stock

stock_per_category_xml
```
