**3. Collect dataset (Customer & Order tables) and apply basic data preprocessing**

```python
df_customers = pd.read_csv('customers.csv')
df_orders = pd.read_csv('orders.csv')
df_customers
```

```python
df_orders
```

```python
df_customers.isnull().sum()
```

```python
df_orders.isnull().sum()
```

```python
df_customers['email'] = df_customers['email'].fillna('Unknown')
df_customers = df_customers.dropna(subset=['join_date'])
df_orders['amount'] = df_orders['amount'].fillna(df_orders['amount'].mean())
df_customers.isnull().sum()
```

```python
df_merged = pd.merge(df_customers, df_orders, on='customer_id', how='inner')
df_merged.head()
```

```python
df_merged['join_date'] = pd.to_datetime(df_merged['join_date'])
df_merged['order_date'] = pd.to_datetime(df_merged['order_date'])
df_merged.dtypes
```
