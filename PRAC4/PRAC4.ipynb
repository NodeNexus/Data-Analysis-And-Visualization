**This cell fetches historical air pollution data for Nagpur.**

```python
def display_air_quality(data):
    aqi_scale = {1: "Good",2: "Fair",3: "Moderate",4: "Poor",5: "Very Poor"}
    aqi_value = data['list'][0]['main']['aqi']
    qualitative_name = aqi_scale.get(aqi_value, "Unknown")
    print(f"Current Air Quality Index (AQI): {aqi_value} - {qualitative_name}")
    print("-" * 40)
    print("Pollutant Concentrations (μg/m3):")
    
    components = data['list'][0]['components']
    for pollutant, concentration in components.items():
        print(f" • {pollutant.upper()}: {concentration}")
display_air_quality(data3)
```

**This cell processes and displays the Air Quality Index and pollutant concentrations.**
